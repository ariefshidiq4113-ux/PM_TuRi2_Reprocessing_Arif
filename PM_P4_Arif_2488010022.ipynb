Langkah 1: Menangani Nilai Hilang (Imputasi)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
data = {
    'usia': [25, np.nan, 35, 45, 22, 30, np.nan, 50],
    'pendapatan': [5000000, 7000000, np.nan, 12000000, 4000000, 8000000, 9500000, np.nan],
    'pendidikan': ['SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'S1', 'SMA'],
    'kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung'],
    'membeli': ['Tidak', 'Ya', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya']
}

df = pd.DataFrame(data)
print("--- Dataset Awal ---")
print(df)
print("\n" + "="*50 + "\n")
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

Langkah 2: Memisahkan Fitur (X) dan Label (y)

```python
# Mengisi nilai hilang pada kolom 'usia' dan 'pendapatan' dengan nilai median
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
X = df.drop(columns=['membeli'])
y = df['membeli']
```

Langkah 3: Encoding Kategorikal

```python
# 3a. Ordinal Encoding untuk tingkat pendidikan
X['pendidikan'] = X['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})

# 3b. One-Hot Encoding untuk variabel nominal 'kota'
X = pd.get_dummies(X, columns=['kota'], dtype=int)

# 3c. Encoding Label/Target 'membeli'
y = y.map({'Tidak': 0, 'Ya': 1})
```

Langkah 4: Membagi Data (SEBELUM Scaling)

```python
# Membagi data menjadi 70% data latih dan 30% data uji
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

Langkah 5: Penskalaan Fitur (Scaling)

```python
# Menentukan kolom numerik yang akan diskalakan
num = ['usia', 'pendapatan', 'pendidikan']

sc = StandardScaler()

# Mencegah data leakage:
# fit_transform pada data latih, transform pada data uji
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])
```

Langkah 6: Verifikasi Hasil

```python
print("--- Hasil X_train setelah Preprocessing & Scaling ---")
print(X_train)
print("\nJumlah nilai hilang di X_train:", X_train.isnull().sum().sum())
print("Tipe data di X_train:\n", X_train.dtypes)

print("\n--- Hasil X_test setelah Preprocessing & Scaling ---")
print(X_test)
print("\nJumlah nilai hilang di X_test:", X_test.isnull().sum().sum())
```

##4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

Penerapan fit_transform hanya pada data latih (X_train) bertujuan untuk menghitung parameter statistik data, seperti rata-rata dan standar deviasi pada StandardScaler atau nilai minimum dan maksimum pada MinMaxScaler sekaligus mentransformasikannya, sedangkan data uji (X_test) hanya menggunakan transform agar proses penskalaan data uji sepenuhnya menggunakan parameter statistik yang telah dipelajari dari data latih tersebut. Hal ini sangat krusial untuk mencegah terjadinya data leakage (kebocoran data), sehingga model dapat dievaluasi secara objektif seolah-olah menghadapi data baru yang belum pernah dilihat sebelumnya tanpa terpengaruh oleh informasi statistik dari data uji

#E. Refleksi

1. Mengapa urutan 'split dulu, baru scaling' penting?

Melakukan pembagian data (train-test split) terlebih dahulu sebelum penskalaan (scaling) sangat penting untuk mencegah terjadinya data leakage atau kebocoran informasi dari data uji ke dalam proses pelatihan model.

2. Kapan label encoding dan kapan one-hot encoding?

Label encoding (atau ordinal encoding) digunakan ketika variabel kategorikal memiliki urutan, tingkatan, atau hierarki yang jelas (data ordinal), seperti tingkat pendidikan (SMA, S1, S2) atau skala kepuasan pelanggan, sehingga representasi angka ($0, 1, 2, ...$) dapat mempertahankan hubungan bertingkat tersebut. Sebaliknya, one-hot encoding digunakan untuk variabel kategorikal nominal yang tidak memiliki tingkatan antar kategorinya, seperti nama kota (Jakarta, Bandung, Surabaya) atau jenis kelamin, guna mencegah algoritma pembelajaran mesin salah mengartikan kategori tersebut sebagai nilai numerik bernilai lebih tinggi atau lebih rendah.

3. Apa perbedaan normalisasi dan standardisasi?

Perbedaan utama antara normalisasi dan standardisasi terletak pada rentang skala dan asumsi distribusi data yang digunakan. Normalisasi (seperti MinMaxScaler) mengubah skala fitur numerik ke dalam rentang terbatas yang pasti, biasanya antara $0$ hingga $1$, sehingga sangat sensitif terhadap data pencilan (outlier). Sementara itu, standardisasi (seperti StandardScaler) menggeser rata-rata (mean) data menjadi $0$ dan standar deviasinya menjadi $1$, sehingga tidak membatasi nilai pada rentang tertentu dan lebih tangguh (robust) terhadap pengaruh outlier serta sangat cocok digunakan pada algoritma yang mengasumsikan data terdistribusi normal.
